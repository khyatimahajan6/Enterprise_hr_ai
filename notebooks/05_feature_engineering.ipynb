# 05 Feature Engineering

Create domain-specific engineered features: income per year at company, promotion gap ratio, and satisfaction index.

```python
import pandas as pd

df = pd.read_csv('../data/processed/employee_attrition_processed.csv')
df['Income_Per_Year'] = df['MonthlyIncome'] / (df['YearsAtCompany'] + 1)
df['Promotion_Gap_Ratio'] = df['YearsSinceLastPromotion'] / (df['YearsAtCompany'] + 1)
df['Satisfaction_Index'] = (df['JobSatisfaction'] + df['EnvironmentSatisfaction'] + df['WorkLifeBalance']) / 3.0
df.head(3)
```
