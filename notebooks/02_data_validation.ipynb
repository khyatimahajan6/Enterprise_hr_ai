# 02 Data Validation

Validate schemas, value ranges, category consistency, and primary key uniqueness.

```python
import pandas as pd

df = pd.read_csv('../data/raw/employee_attrition.csv')
```

```python
assert df['Age'].between(18, 100).all(), 'Age out of valid range'
print('[✓] Age validation passed')
```

```python
assert df['EmployeeNumber'].is_unique, 'Duplicate EmployeeNumber found'
print('[✓] EmployeeNumber uniqueness passed')
```

```python
assert set(df['Attrition'].unique()) <= {'Yes', 'No'}, 'Unexpected Attrition values'
print('[✓] Attrition categories passed')
```
